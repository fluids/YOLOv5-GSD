## 1. Download and Extract the Data

Download files with names beginning with `NvidiaH100_` from the [dataset](https://doi.org/10.15485/2588483). Extract the files and move all extracted folders into a directory of your choice. ## 2. Hardware and Model

These examples were tested on an NVIDIA H100 GPU. Other compatible NVIDIA GPUs may also be used with an appropriate CUDA-enabled PyTorch installation.

The YOLOv5 model used here is the **M0c model** described in:

Chen, Y., et al. (2024). “Quantifying streambed grain size, uncertainty, and hydrobiogeochemical parameters using machine learning model YOLO.” *Water Resources Research*, 60, e2023WR036456. https://doi.org/10.1029/2023WR036456

## 5. Before Running

Check the `WorkDir` and `overwrite` settings. Set `WorkDir` to the directory containing your extracted data. Setting `overwrite=True` will overwrite existing output files.

## Run the code

#### Importing packages

```python
import os, sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
import AI4GSD
```

#### Define data directory.

```python
WorkDir = Path('/rcfs/projects/rcsfa/VisionAI')
folders = [
    'NvidiaH100_NoCut_12Filter',
    'NvidiaH100_NoCut_12Filter_Run2',
    'NvidiaH100_NoCut_12Filter_Run3',
    'NvidiaH100_0.1Cut_12Filter',
    'NvidiaH100_NoCut_12Filter_SingleDiameter',
]
keyword = '20260430_MairD2026IG2'
overwrite = True
save_overlay = True
device = '0'
hide_conf = True
```

#### Call YOLOv5 to process all data.

```python
for master_folder in folders:
    data_dir = WorkDir / master_folder
    subfolders = AI4GSD.getFolders(str(data_dir))
    print(f'Master folder: {master_folder}')
    
    for subfolder in subfolders:
        img_size = 320 if keyword.lower() in subfolder.lower() else 1280
        photo_dir = data_dir / subfolder
        AI4GSD.yolov5_predict(
            img_dir=str(photo_dir),
            img_size=img_size,
            overwrite=overwrite,
            device=device,
            save_overlay=save_overlay,
            line_thickness=2,
            hide_conf=hide_conf,
        )
    print('')
```
